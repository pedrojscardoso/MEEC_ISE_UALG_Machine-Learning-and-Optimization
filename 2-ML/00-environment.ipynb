# Configuring the Python environment

This notebook explains how to set up a **dedicated Python environment** for the classes. You only need to do this once per machine.

An isolated environment keeps the packages used in the course separate from other Python projects on your computer, and makes it easier to reproduce the same setup later.

**Choose one of the two alternatives below** — do not mix them:

1. **Anaconda** — a full scientific-Python distribution, with a graphical installer. A good default if you prefer a GUI.
2. **uv** — a fast, modern package and Python manager. A good default if you are comfortable with the terminal.

Both options install the same libraries and can run the same notebooks.

With **Anaconda** you can do the setup in **Anaconda Navigator** (GUI) or in a terminal. With **uv**, the setup is done in a terminal. Those commands are meant to be run in a **terminal** (macOS/Linux) or **Anaconda Prompt** / **PowerShell** (Windows), not inside this notebook.

## What we will install

We recommend **Python 3.12** and the following packages, which cover the Python review and most of the Machine Learning notebooks:

| Package | Role |
| --- | --- |
| `numpy` | arrays and numerical computing |
| `pandas` | tables and data frames |
| `matplotlib` | plots |
| `scikit-learn` | machine-learning algorithms |
| `jupyterlab` | notebooks |
| `ipykernel` | Jupyter kernel for this environment |
| `joblib` | saving and loading models |

A few later notebooks (neural networks with Keras/TensorFlow, text processing) need extra packages. Those are listed at the end.

## 2. Create the course environment

Use **either** Anaconda Navigator **or** the terminal — not both for the same environment.

### 2a. Anaconda Navigator (GUI)

This path requires the **full Anaconda Distribution** (Navigator is not included with Miniconda).

1. Open **Anaconda Navigator**.
2. Click **Environments** in the left sidebar.
3. Click **Create** at the bottom of the environment list.
4. In the dialog:
   - **Name:** `mlo`
   - **Packages:** Python
   - **Version:** 3.12 (or the closest 3.12.x listed)
5. Click **Create** and wait until `mlo` appears in the list.

Install the course packages:

6. Select **mlo** in the environment list.
7. Open the package filter dropdown and choose **Not installed**.
8. Search for and tick: `numpy`, `pandas`, `matplotlib`, `scikit-learn`, `jupyterlab`, `ipykernel`, `joblib`.
9. Click **Apply**, review the list, then click **Apply** again.

If a package does not appear, click **Channels**, add `conda-forge`, and search again.

Launch JupyterLab from Navigator:

10. Go back to the **Home** tab.
11. At the top, set *Applications on* to **mlo**.
12. Click **Launch** under **JupyterLab** (click **Install** first if that is what the button shows).

In the notebook menu, choose **Kernel → Change Kernel** and select the kernel that corresponds to `mlo`.

### 2b. Terminal (`conda`)

```bash
conda create -n mlo python=3.12
conda activate mlo
```

Install the core packages. The `conda-forge` channel is well maintained and is the recommended source:

```bash
conda install -c conda-forge numpy pandas matplotlib scikit-learn jupyterlab ipykernel joblib
```

Confirm that the environment is active: the prompt should start with `(mlo)`.

## 3. Register a Jupyter kernel and start the notebooks (terminal only)

If you launched JupyterLab from Navigator, skip this section.

```bash
python -m ipykernel install --user --name mlo --display-name "Python (mlo)"

```

In the notebook menu, choose **Kernel → Change Kernel → Python (mlo)**.

To work later: open a terminal, run `conda activate mlo`, then `jupyter lab`. In VS Code, Cursor, or PyCharm, select the `mlo` interpreter instead (see [Using the environment in an IDE](#using-the-environment-in-an-ide)).

# Alternative B — uv

[uv](https://docs.astral.sh/uv/) is a fast Python package manager from Astral. It can also **install Python itself**, so you do not need Anaconda if you follow this path.

## 1. Install uv

**macOS / Linux:**

```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

**Windows (PowerShell):**

```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

Alternatively, with Homebrew: `brew install uv`.

Close and reopen the terminal, then check:

```bash
uv --version
```

## 2. Create a virtual environment

From the **root of this repository** (the folder that contains `1-Python` and `2-ML`):

```bash
uv python install 3.12
uv venv --python 3.12
```

This creates a `.venv` directory. Do not commit it to git.

Activate the environment:

**macOS / Linux**

```bash
source .venv/bin/activate
```

**Windows (Command Prompt)**

```bat
.venv\Scripts\activate
```

**Windows (PowerShell)**

```powershell
.venv\Scripts\Activate.ps1
```

Install the core packages:

```bash
uv pip install numpy pandas matplotlib scikit-learn jupyterlab ipykernel joblib
```

You can skip activation and prefix commands with `uv run` instead, for example `uv run jupyter lab`.

## 3. Register a Jupyter kernel and start the notebooks

```bash
uv run python -m ipykernel install --user --name mlo --display-name "Python (mlo)"
uv run jupyter lab
```

In the notebook menu, choose **Kernel → Change Kernel → Python (mlo)**.

# Troubleshooting

- **`conda` / `uv` not found** — close and reopen the terminal (or Anaconda Prompt). On macOS/Linux you may need a new shell so that `PATH` is updated.
- **Wrong kernel** — the notebook is running another Python. Change the kernel to **Python (mlo)** and re-run the check cell.
- **Do not mix tools** — pick Anaconda *or* uv. Installing the same packages with both in overlapping ways is a common source of confusion.
- **Permission / policy errors on Windows** — if PowerShell blocks activation scripts, run `Set-ExecutionPolicy -Scope CurrentUser RemoteSigned`.
- **Apple Silicon (M1/M2/M3/M4)** — both conda-forge and uv provide native `osx-arm64` builds for the core packages. Prefer those over x86 compatibility layers.
