In this exercise, we will use the well-known diabetes dataset, which contains data collected from 442 patients, each described with 10 physiological variables (such as age, sex, BMI, blood pressure, and various blood serum measurements).

The objective is to build a predictive model to estimate the progression of diabetes in these patients, quantified as a disease progression measure one year after baseline measurements.

**What is being predicted.** There is no sequence of visits in this table. Each row is one patient. The ten columns are a single baseline snapshot (age, sex, BMI, blood pressure, and six serum measurements). The target is one number recorded **one year later**: how far the disease has progressed since that baseline. A larger value means more progression. In this file the scores run from 25 to 346. It is a regression target, not a class such as "diabetic" or "not diabetic".

**How it is measured.** The published dataset, from Efron et al. (2004), calls this a quantitative measure of disease progression one year after baseline. It does not name the clinical instrument or the units, and the score is **not** computed from the ten baseline features. Those features are the inputs. The one-year score is the quantity the model must predict.
