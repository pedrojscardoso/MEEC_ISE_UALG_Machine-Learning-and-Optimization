# Learning and predicting

In the digits dataset the task is to **predict which digit (0–9) an 8×8 image represents**.

We are given labelled examples of each class. We **fit an estimator** on those examples and then use it to **predict the class of unseen images**.

In scikit-learn, a classifier is an object that implements two methods:

- `fit(X, y)` — learn from the training data
- `predict(X)` — assign a class to new instances

This notebook is a first, very simple walk-through. We treat the model as a **black box**: we will not discuss support vector machines yet.

```python
%matplotlib inline

from sklearn import datasets, svm
import matplotlib.pyplot as plt
```

```python
digits = datasets.load_digits()
digits.data.shape
```

## Create a classifier

`sklearn.svm.SVC` implements support vector classification. The constructor receives the model hyperparameters (`gamma`, `C`, …).

**For now we treat the estimator as a black box** — the values below are simply a working example.

```python
clf = svm.SVC(gamma=0.001, C=100.)
```

## Fit the model (learn from data)

We call the estimator `clf` because it is a classifier.

To *learn*, we pass a training set to `fit`. Here we use **all images except the last five** as training data, selected with the Python slice `[:-5]`.

This is only a quick demonstration. Holding out a few samples at the end of the array is **not** a proper train/test split; we will see better ways later (random splits, cross-validation).

```python
training_set = digits.data[:-5]
target_set = digits.target[:-5]
```

scikit-learn estimators follow a common convention: they expose a `fit` method that trains the model on `X` (features) and `y` (labels).

```python
clf.fit(training_set, target_set)
```

## Predict

What does the classifier predict for the five images we left out?

```python
clf.predict(digits.data[-5:])
```

The true labels of those five images are:

```python
digits.target[-5:]
```

Do the predictions match the true labels? Let us look at the images themselves.

```python
m, n = digits.data.shape
for image in digits.data[-5:]:
    image = image.reshape((8,8))
    print(image)
```

## Visualize the held-out images

The raw 8×8 matrices are hard to read. Plotting them with `imshow` makes the digits easier to recognize.

```python
fig, ax = plt.subplots(1, 5, figsize=(25, 5))
for i, image in enumerate(digits.data[-5:]):
    image = image.reshape((8,8))
    ax[i].imshow(image, cmap=plt.cm.gray_r, interpolation='nearest')
plt.show()

# to recall
print(digits.target[m-5:])
```
